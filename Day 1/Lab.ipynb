```python
# 5. Let x = np.arange(1, 21).reshape(4, 5)
# (a) The last two rows
# (b) The third column
# (c) The 3×3 submatrix in the top-right corner

x = np.arange(1, 21).reshape(4, 5)
print(x)

# (a) The last two rows
print(x[-2:,:])

# (b) The third column
third_col=x[:,2]

# (c) The 3×3 submatrix in the top-right corner
submatrix = x[:3,2:]
print(submatrix)
```

```python
# 7. Normalize the array to range [0, 1]
data = np.array([150, 220, 98, 360, 180, 45])

normalized_data = (data - data.min())/ (data.max() - data.min())
print(normalized_data)
```

```python
# 8. Standardize (z-score) the same array from Question 7
mean = data.mean()
std = data.std()

standardized_data = (data - mean) / std
print(standardized_data)
```

```python
# 9. Given the 2D array
    
mat = np.array([
    [4, 8, 1, 12],
    [7, 3, 9, 5],
    [11, 2, 6, 10],
    [15, 0, 14, 13]
])

# a. Compute the mean of each row
mat_row_mean = mat.mean(axis=1)

# b. Compute the sum of each column
mat_col_sum = mat.sum(axis= 0)

# c. Find the maximum value in the matrix and its position
mat_max_v = mat.max()

max_val_pos = np.unravel_index(mat.argmax(), mat.shape)
print(mat)
print(mat_row_mean)
print(mat_col_sum)
print(mat_max_v)
print(max_val_pos)
```
